# Notebook 02: Data Preprocessing & YOLO Formatting

This notebook splits annotated data into `train/`, `val/`, and `test/` splits, validates normalized bounding box coordinates (`[0.0, 1.0]`), and generates dataset splits.

```python
import os
import shutil
import random
from pathlib import Path

BASE_DIR = Path("..").resolve()
DATASET_DIR = BASE_DIR / "dataset"
ANNOTATED_IMG = DATASET_DIR / "annotated" / "images"
ANNOTATED_LBL = DATASET_DIR / "annotated" / "labels"

print(f"Source annotated images: {ANNOTATED_IMG}")
```

```python
def split_dataset(train_ratio=0.7, val_ratio=0.2, test_ratio=0.1, seed=42):
    random.seed(seed)
    images = list(ANNOTATED_IMG.glob("*.*"))
    images = [img for img in images if img.suffix.lower() in [".jpg", ".jpeg", ".png"]]
    
    if not images:
        print("No annotated images found yet. Place your labeled images in dataset/annotated/")
        return
        
    random.shuffle(images)
    n = len(images)
    n_train = int(n * train_ratio)
    n_val = int(n * val_ratio)
    
    train_imgs = images[:n_train]
    val_imgs = images[n_train:n_train+n_val]
    test_imgs = images[n_train+n_val:]
    
    print(f"Total: {n} | Train: {len(train_imgs)} | Val: {len(val_imgs)} | Test: {len(test_imgs)}")

split_dataset()
```
