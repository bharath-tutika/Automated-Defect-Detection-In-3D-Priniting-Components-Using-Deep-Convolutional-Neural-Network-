# Notebook 01: Dataset Analysis & Class Distribution

This notebook inspects the 3D printing defect dataset, checks class balance across the 7 supported classes, verifies image dimensions, and inspects bounding box statistics.

```python
import os
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
import yaml

# Set base project path
BASE_DIR = Path("..").resolve()
DATASET_DIR = BASE_DIR / "dataset"
DATA_YAML = DATASET_DIR / "data.yaml"

with open(DATA_YAML, "r") as f:
    dataset_config = yaml.safe_load(f)

print("Classes in data.yaml:")
for k, v in dataset_config["names"].items():
    print(f"  Class {k}: {v}")
```

```python
# Inspect Raw Dataset Subdirectories
raw_dir = DATASET_DIR / "raw"
class_counts = {}

for cls_folder in raw_dir.iterdir():
    if cls_folder.is_dir():
        images = [f for f in cls_folder.glob("*") if f.suffix.lower() in [".jpg", ".jpeg", ".png", ".webp"]]
        class_counts[cls_folder.name] = len(images)

df_counts = pd.DataFrame(list(class_counts.items()), columns=["Class", "Image Count"])
print(df_counts)

# Plot distribution
plt.figure(figsize=(10, 5))
plt.bar(df_counts["Class"], df_counts["Image Count"], color="#3b82f6")
plt.title("3D Printing Defect Dataset - Class Distribution")
plt.xlabel("Defect Category")
plt.ylabel("Number of Images")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
