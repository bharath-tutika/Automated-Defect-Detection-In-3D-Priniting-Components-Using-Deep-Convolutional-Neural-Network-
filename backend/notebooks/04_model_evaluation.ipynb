# Notebook 04: Model Evaluation & Metric Extraction

This notebook evaluates trained YOLO weights against the validation & test sets, calculating Precision, Recall, F1-Score, and Mean Average Precision (mAP@50 and mAP@50-95).

```python
from pathlib import Path
from ultralytics import YOLO

BASE_DIR = Path("..").resolve()
MODEL_PATH = BASE_DIR / "models" / "trained" / "best.pt"
DATA_YAML = str(BASE_DIR / "dataset" / "data.yaml")

if MODEL_PATH.exists():
    model = YOLO(str(MODEL_PATH))
    metrics = model.val(data=DATA_YAML, split="val")
    print(f"Precision: {metrics.box.mp:.4f}")
    print(f"Recall:    {metrics.box.mr:.4f}")
    print(f"mAP@50:    {metrics.box.map50:.4f}")
    print(f"mAP@50-95: {metrics.box.map:.4f}")
else:
    print(f"Model weights not found at {MODEL_PATH}. Train a model first.")
```
